# load sample

```python
import pandas as pd
from pathlib import Path
import json

# get sample ids
sample_path = Path("../../data/sample/sample_tuning_10per.json")

with open(sample_path, "r") as f:
    sample_info = json.load(f)

#sample_ids = pd.DataFrame(sample_info["sample"])
sample_info.keys()
```

```python
#sample_info["sampling_by_stratum"]
sampling = pd.DataFrame(sample_info["sampling_by_stratum"])

sampling
```

```python
table = sampling.pivot(
    index="dominant_field",
    columns="text_type",
    values=["sample", "percentage"]
)

def format_cell(sample, percentage):
    if pd.isna(sample) or pd.isna(percentage):
        return "--"
    return f"{sample:,.0f} ({percentage:.1f}\\%)"

formatted = pd.DataFrame(index=table.index)

for text_type in ["abstract", "title"]:
    formatted[text_type] = [
        format_cell(sample, percentage)
        for sample, percentage in zip(
            table[("sample", text_type)],
            table[("percentage", text_type)]
        )
    ]

formatted = formatted.rename(columns={
    "abstract": "Abstract",
    "title": "Title"
})
```

```python
formatted
```

```python
latex = formatted.to_latex(
    escape=False,
    index=True,
    column_format="lrr",
    header=True
)

print(latex)
```
